```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.model_selection import train_test_split

data = {
    "studyhrs" : [3, 4, 5, 6, 7],
    "score" : [40, 50, 60, 70, 80]

}

df = pd.DataFrame(data)

#standardscaler

scaler = StandardScaler()

standerd_scaled = scaler.fit_transform(df)

print("Scaled output")
print(pd.DataFrame(standerd_scaled, columns=["studyhrs", "score"]))

"""
the value of standard scaler = z=x-mew/standard_deviation """


mm_scaler = MinMaxScaler()
mm_scaled = mm_scaler.fit_transform(df)

print("MM_scaled output : ")
print(pd.DataFrame(mm_scaled, columns=["studyhrs", "score"]))

#minmax_scaler is used to scale the values in such a way that they lie in between 0 and 1 
#formula is scaled = x - xmin/xmax-xmin

x = df[["studyhrs"]]
y = df[["score"]]

x_train, x_test, y_trian, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
print("Training data :")
print(x_train)

print("Test_data : ")
print(x_test)

#here we are splitting the testing and training ratio so that it learns from the trained data and then practices its accuracy on the testing data 
```

```python
'''
linear Regression 
1- finds pattern in old data 
2- straight line 

y = m*x + c
'''

from sklearn.linear_model import LinearRegression


X = [[1], [2], [3], [4], [5]]
Y = [20, 35, 40, 50, 70]
model = LinearRegression()

model.fit(X, Y)

hours = float(input("Enter the no of hours you studied : "))
p_marks = model.predict([[hours]])

print(f"Based on your studies hours you will get {p_marks} marks")

```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()

X = [[2], [3], [4], [5], [7]] #hours studied 
Y = [0, 0, 0, 1, 1] #pass or fail 0-f, 1-p

model.fit(X, Y)

hours = float(input("Enter the number of hours studied : "))

result = model.predict([[hours]])[0]

#print(f"Based on the hours studied you might {result} ")

if result == 1:
    print(f"Based on the hours studied (which is {hours}) you might pass.")
else:
    print("Based on the hours studied {hours} you might fail.")
```

```python
'''#KNN K NEAREST NEIGHBOURS
#creates a decision by checking out the neighbours 
#works very slow for big data 
#always put negative value of k 
# '''


from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier(n_neighbors=3)
X = [
    [180, 7],
    [200, 7.5],
    [250, 8],
    [300, 8.5],
    [330, 9],
    [360,9.5]
]
Y = [0,0,0,1,1,1,] #0 is apple , 1 is orange
model.fit(X,Y)
weight = float(input("Enter the weight in Grams : "))
size = float(input("Enter the size in cms : "))

prediction = model.predict([[weight, size]])[0]

if prediction ==1:
    print("its an orange")
else:
    print("Its an apple.")
```

```python
'''
Is Fever > 100
    yes > Cough?
        yes > flu
        No > Maybe
    No - No Flu
'''

from sklearn.tree import DecisionTreeClassifier

X = [[7, 2],
     [8, 3],
     [9, 8],
     [10, 9]
]
Y  = [0,0,1,1]  #0 - apple, 1 - orange
model = DecisionTreeClassifier()

model.fit(X, Y)
size = float(input("Enter the size in cms : "))
shade  = float(input("Enter the color shade (1-10) : "))

predicted_fruit = model.predict([[size, shade]])[0]
if predicted_fruit ==0:
    print("Fruit is likely to be an apple.")
else:
    print("Fruit is likely to be an orange.")
```
