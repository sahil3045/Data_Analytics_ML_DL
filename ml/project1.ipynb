```python
import numpy as np 
import pandas as pd
from sklearn.metrics import mean_absolute_error,mean_squared_error
from sklearn.linear_model import LinearRegression

data = pd.read_csv("student.csv")

x = data[["Hours"]]

y = data["Score"]

model = LinearRegression()
model.fit(x,y)

predicted_score = model.predict(x)

mse = mean_squared_error(y, predicted_score)
mae = mean_absolute_error(y, predicted_score)
rmse = np.sqrt(mse)

print("Mean abosulte error : ", mae)
print("Mean Squared Error : ", mse)
print("Root Mean Squared Error : ", rmse)

pred = float(input("Enter the no of hours studied : "))
new_pred = model.predict([[pred]])
print(f"Prediction for new hour : {pred} is {new_pred}.")
```

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

data = pd.read_csv("student_performance.csv")

x = data[["weekly_self_study_hours"]]
y = data["total_score"]

model = LinearRegression()
model.fit(x,y)

predicted_score = model.predict(x)

mse = mean_squared_error(y, predicted_score)
mae = mean_absolute_error(y, predicted_score)
rmse = np.sqrt(mse)
r2_s = r2_score(y, predicted_score)

'''
round(x,2) it basically rounds up values up to 2 decimal digit
'''


print("Mean abosulte error : ", round(mae,2))
print("Mean Squared Error : ", round(mse,2))
print("Root Mean Squared Error : ", round(rmse,2))
print("R2_score : ", round(r2_s,2)) #closer to 1 its better , lower than 0 than model is worst 

#histogram 
plt.figure(figsize=(10, 6)) #width, height 
plt.hist(data["total_score"], bins = 30, color="skyblue")
plt.title("Distribution of final exam score.")
plt.xlabel("Final Exam Score.")
plt.ylabel("No of students ")
plt.grid(True)
plt.show()


#scatter plot + regression line 
plt.figure(figsize=(10, 5)) #width, height 
plt.scatter(x, y, color="blue", label = "Actual score")
plt.plot (x, predicted_score, color="red", label="Predicted Scores (Regression Line)")
plt.title("Model Prediction vs actual scores.")
plt.xlabel("Study hours per week.")
plt.ylabel("Final Score.")
plt.grid(True)
plt.show()

```
