```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
#True answers data 
#what actually happened
y_true = [1, 0, 1, 1, 0, 1, 0]
#what our model guessed
y_pred = [1, 0, 1, 0, 0, 1, 1]

print("Accuracy : ", accuracy_score(y_true, y_pred))
print("Precision : ", precision_score(y_true, y_pred))
print("Recall : ", recall_score(y_true, y_pred))
print("F1_Score : ", f1_score(y_true, y_pred))
```

```python
from sklearn.metrics import confusion_matrix
y_true = [1, 0, 0, 1, 0, 1, 0, 1, 1, 1, 0]
y_pred = [1, 0, 1, 1, 1, 0, 0, 1, 1, 0, 1]

cm = confusion_matrix(y_true, y_pred )
print("confusion matrix", cm)
```

```python
'''
MAE - Mean Absolute Error,
take mistake difference -> remove minus sign -> add -> divide 

MSE - Mean Squared Error 
square the mistake -> add -> divide
to detect the outliers 

RMSE - Root mean squared error

'''
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np 

real_score = [90, 60, 80, 100]
predicted_score = [85, 70, 70, 95]

mse = mean_squared_error(real_score, predicted_score)
mae = mean_absolute_error(real_score, predicted_score)

rmse = np.sqrt(mse)
print(mse,mae,rmse, end = " ")
```
